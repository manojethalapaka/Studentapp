```python
import matplotlib.pyplot as pp 
import numpy as np 
x=np.array([1,2,3,4,5])
y=np.array([6,7,8,9,10])
z=np.array([7,4,2,3,7])
pp.plot(x,y,marker="d",color="red",mec='black',mfc='blue',label='1st-line',linestyle='--',linewidth=1.5)
pp.plot(x,z,marker="s",mec='red',mfc='k',label='2nd-line')
pp.title("LINE GRAPH",color="k")
pp.xlabel("X - axis",color="k")
pp.ylabel("Y - axis",color="k")
pp.legend(title="GRAPH DETAILS",loc="upper left")
pp.show()

```

```python
pp.bar(x,y,color=["red","green","blue","k","orange"],width=0.7,alpha=0.7)
pp.title("BAR GRAPH",color="maroon")
pp.xlabel("X-AXIS",color="k")
pp.ylabel("Y-AXIS",color="k")
pp.show()
```

```python
pp.barh(x,y,color=["orange","magenta","blue","maroon","purple"],height=0.7,alpha=0.8)
pp.title("BARH GRAPH",color="maroon")
pp.xlabel("X-AXIS",color="k")
pp.ylabel("Y-AXIS",color="k")
pp.show()
```

```python
fields=["OnePiece","Naruto","DemonSlayer","AttackOnTitan","SoloLevelling"]
values=[30,23,20,17,10]
pp.pie(values,labels=fields,autopct="%1.1f%%")
pp.title("PIE CHART",color='k')
pp.show()
```

```python
#view student details
def view_students_info():
    return(df)
view_students_info()
```

```python
def add_student():
    student_id=input('enter student id :')
    name=input("enter student name:").title()
    department=input("enter student department:").upper()
    year=int(input("enter student academic year:"))
    email=input("enter student email:")
    phone=int(input("enter student phone no:"))
    
    df.loc[len(df)]=[student_id,name,department,year,email,phone]
    
    print("student added successfully")
    
add_student()
```

```python
def delete_student():
    global df
    student_id=input("enter student id:").upper()
    df=df[df['Student_ID']!=student_id]
    print("student data deleted successfully")

delete_student()
    
```

```python
view_students_info()
```

```python
#updating student data
def update_student():
    global df
    student_id=input("enter student id:").upper()
    
    index=df[df['Student_ID']==student_id].index
    
    name=input("enter student name:").title()
    department=input("enter student department:").upper
    year=int(input("enter student academic year:"))
    email=input("enter student email:")
    phone_no=int(input("enter student phone no:"))
    
    df.loc[index,'Name']=name
    df.loc[index,"Department"]=department
    df.loc[index,'Year']=year
    df.loc[index,'Email']=email
    df.loc[index,'Phone']=phone_no
    
    print("student details updated successfully")
    
update_student()
```
